```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import sklearn
```

```python
df = pd.read_csv('../datasets/car_price_prediction_.csv', index_col='Car ID')
df.head()
```

```python
print(df.info())
# DataFrame haven't null cells

print('-.-'*30)
print(df.describe())

print('-.-'*30)
print(df.isna().sum())
```

```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(14, 10))

sns.histplot(data=df, x="Price", kde=True, ax=axes[0, 0], color='skyblue')
axes[0, 0].set_title('Distribution of car prices')

sns.scatterplot(data=df, x='Mileage', y='Price', ax=axes[0, 1], alpha=0.6, color='salmon')
axes[0, 1].set_title('Price vs Mileage')

sns.boxplot(data=df, x='Fuel Type', y='Price', ax=axes[1, 0], palette='pastel')
axes[1, 1].set_title('Price by fuel type')

sns.countplot(data=df, x='Year', ax=axes[1, 1], palette='viridis')
axes[1, 1].set_title('Nubmer vehicle by year')
axes[1, 1].tick_params(axis='x', rotation=45)

plt.tight_layout
plt.show()
```
